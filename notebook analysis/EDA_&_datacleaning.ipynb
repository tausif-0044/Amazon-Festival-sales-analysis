```python
#DATA CLEANING----------------------------------------------------------------
```

```python
#copying data for data cleaning
df_cleaned = df.copy()
```

```python
#CLEANING THE COLUMNS
df_cleaned.columns = (
    df_cleaned.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[\s-]+", "_", regex=True)
)
```

```python
#STANDARISING COLUMNS
text_columns = df_cleaned.select_dtypes(
    include=["object", "string"]
).columns

for col in text_columns:
    df_cleaned[col] = df_cleaned[col].str.strip()

df_cleaned.head(1)
```

```python
#STANDARISING COLUMN VALUES

df_cleaned["ship_city"] = df_cleaned["ship_city"].str.upper()
df_cleaned["ship_state"] = df_cleaned["ship_state"].str.upper()
df_cleaned["category"] = df_cleaned["category"].str.title()
df_cleaned.head(1)
```

```python
#deleting duplicates data
df_cleaned = df_cleaned.drop_duplicates()
```

```python
#DELETING unneccessary COLUMNS
df_cleaned = df_cleaned.drop(
    columns = ["promotion_ids", "unnamed:_22"]
)
df_cleaned.head(1)
```

```python
df_cleaned.isnull().sum()
```

```python
#changing data type object to string
df_cleaned["ship_city"] = df_cleaned["ship_city"].astype(str)
```

```python
#displaying null values rows
display(df_cleaned[df_cleaned["ship_city"].isna()])
```

```python
#replacing ship_city and ship_state to Unknown
df_cleaned[["ship_city", "ship_state"]] = df_cleaned[
    ["ship_city", "ship_state"]
].fillna("Unknown")
```

```python
#filling null values in amount column with mean value
df_cleaned["amount"] = df_cleaned["amount"].fillna(df_cleaned["amount"].mean())

#filling null values in currency column with INR 
df_cleaned["currency"] = df_cleaned["currency"].fillna("INR")

#filling null values country with IN
df_cleaned["ship_country"] = df_cleaned["ship_country"].fillna("IN")
```

```python
df_cleaned["courier_status"].value_counts()
```

```python
#null count in courier status column
df_cleaned["courier_status"].isnull().sum()
```

```python
#analysing courier status null values
display(df_cleaned[df_cleaned["courier_status"].isna()].head(5))
```

```python
#extracting unique values from status column
statuses = []
print("before extracting : ",statuses)
for i in df_cleaned["status"].unique():
    statuses.append(i)
print("after extracting : ",statuses)
```

```python
#checking relation between courier status and status column

display(
    df_cleaned.loc[
        (df_cleaned["courier_status"].isna()) &
        (df_cleaned["status"].isin(statuses))
    ]
)
```

```python
#replacing courier status column null values with according to status column
for status in statuses:
    mask = df_cleaned["courier_status"].isna() & df_cleaned["status"].eq(status)
    df_cleaned.loc[mask, "courier_status"] = status
print("Null values count in courier_status after replacing NaN values")
df_cleaned["courier_status"].isnull().sum()
```
